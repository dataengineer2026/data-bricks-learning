```python
df_products_1 = spark.read.parquet('/Volumes/ecommerce/e_comm_bronze/e_comm_source/ProductsFile1.parquet')
df_products_2 = spark.read.parquet('/Volumes/ecommerce/e_comm_bronze/e_comm_source/ProductsFile2.parquet')
df_products_1.createOrReplaceTempView('df_products_1')
df_products_2.createOrReplaceTempView('df_products_2')
#df_products = df_products_1.union(df_products_2)
#df_products.createOrReplaceTempView('df_products')
```

```python
df_products = spark.sql("""
        select * from df_products_1
        union all
        select * from df_products_2                        
                        """)
```

```python
df_products.write.format("delta").mode("overwrite").saveAsTable("tblProducts")
```
