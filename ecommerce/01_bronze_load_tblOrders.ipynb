```python
%sql
use catalog ecommerce;
use schema e_comm_bronze;
```

```python
df_orders_1 = spark.read.parquet("/Volumes/ecommerce/e_comm_bronze/e_comm_source/OrdersFile1.parquet")
df_orders_2 = spark.read.parquet("/Volumes/ecommerce/e_comm_bronze/e_comm_source/OrdersFile2.parquet")
df_orders_1.createOrReplaceTempView("orders_01")
df_orders_2.createOrReplaceTempView("orders_02")
```

```python
df_orders = spark.sql("""
        select * from orders_01
        union all
        select * from orders_02
                      """)
```

```python
df_orders.write.format("delta").mode("overwrite").saveAsTable("tblOrders")
```
