```python
%sql
use catalog ecommerce;
use schema e_comm_bronze;
```

```python
df_customers_1= spark.read.parquet('/Volumes/ecommerce/e_comm_bronze/e_comm_source/CustomersFile1.parquet')
df_customers_1.createOrReplaceTempView("customers_01")
df_customers_2= spark.read.parquet('/Volumes/ecommerce/e_comm_bronze/e_comm_source/CustomersFile2.parquet')
df_customers_1.createOrReplaceTempView("customers_02")
```

```python
df_customer = spark.sql("""
            select * from  customers_01
            UNION ALL
            select * from customers_02
                        """)
```

```python
df_customer.write.format("delta").mode("overwrite").saveAsTable("tblCustomers")
```
