#  **Dividing New Records vs Old Records**

```python
%sql
CREATE table DimRegions
(
  DimRegionKey BIGINT GENERATED ALWAYS AS IDENTITY,
  region_id string,
  region string,
  create_date date,    
  update_date date
)
```

## **SCD Type - 1**

```python
from delta.tables import DeltaTable
```

```python
%sql
Merge INTO DimRegions
Using tblRegionsSilver 
ON DimRegions.region_id = tblRegionsSilver.region_id
WHEN MATCHED THEN
UPDATE SET    
  DimRegions.region = tblRegionsSilver.region, DimRegions.update_date = current_date()
WHEN NOT MATCHED THEN
INSERT(region_id, region, create_date) VALUES 
  (tblRegionsSilver.region_id, tblRegionsSilver.region, current_date())
;
```
