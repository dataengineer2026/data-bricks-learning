```python
from pyspark.sql.functions import * 
from pyspark.sql.types import *
import time
```

```python
%sql
use catalog projectv1
```

```python
%sql
use schema projectschema
```

### **Data Reading**

```python
df = spark.sql("select * from tblcustomerssource")
df.display()
```

```python
df = df.dropna()
```

```python
df = df.withColumn("domains",split(col('email'),'@')[1])
df.display()
```

```python
df_gmail = df.filter(col('domains')=="gmail.com")
df_gmail.display()
time.sleep(5)

df_yahoo = df.filter(col('domains')=="yahoo.com")
df_yahoo.display()
time.sleep(5)

df_hotmail = df.filter(col('domains')=="hotmail.com")
df_hotmail.display()
time.sleep(5)
```

```python
df = df.withColumn("full_name",concat(col('first_name'),lit(' '),col('last_name')))
df = df.drop('first_name','last_name')
df.display()
```

```python
df.write.mode("overwrite").format("delta").saveAsTable("tblCustomersSilver")
```
