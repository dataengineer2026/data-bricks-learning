### **Functions**

```python
%sql
CREATE OR REPLACE FUNCTION projectv1.projectschema.discount_func(p_price DOUBLE) 
RETURNS DOUBLE  
LANGUAGE SQL 
RETURN p_price * 0.90
```

```python
%sql
select product_id, upper(brand) as ProductName, category as ProductCategory, price, projectv1.projectschema.discount_func(price) as discounted_price FROM tblProductssource
```

```python
dfnew = spark.sql(""" 
select product_id, upper(brand) as ProductName, category as ProductCategory, price, projectv1.projectschema.discount_func(price) as discounted_price FROM tblProductssource""")
```

```python
dfnew.write.format("delta").saveAsTable("tblProductsSilver")
```
